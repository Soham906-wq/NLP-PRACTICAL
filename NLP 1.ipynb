```python
import nltk
from nltk.tokenize import WhitespaceTokenizer, wordpunct_tokenize, TreebankWordTokenizer, TweetTokenizer, MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = "NLTK is a powerful library! It helps in tokenization, stemming, and lemmatization. Let's test it :) #NLP"

print("Original Text:\n", text)

ws_tokenizer = WhitespaceTokenizer()
ws_tokens = ws_tokenizer.tokenize(text)
print("\nWhitespace Tokenization:\n", ws_tokens)
```

```python
punct_tokens = wordpunct_tokenize(text)
print("\nPunctuation-based Tokenization:\n", punct_tokens)
```

```python
treebank_tokenizer = TreebankWordTokenizer()
treebank_tokens = treebank_tokenizer.tokenize(text)
print("\nTreebank Tokenization:\n", treebank_tokens)
```

```python
tweet_tokenizer = TweetTokenizer()
tweet_tokens = tweet_tokenizer.tokenize(text)
print("\nTweet Tokenization:\n", tweet_tokens)
```

```python
mwe_tokenizer = MWETokenizer([('NLTK', 'is'), ('powerful', 'library')])
mwe_tokens = mwe_tokenizer.tokenize(text.split())
print("\nMWE Tokenization:\n", mwe_tokens)
```

```python

porter = PorterStemmer()
snowball = SnowballStemmer("english")

print("\nPorter Stemming:")
print([porter.stem(word) for word in ws_tokens])
```

```python
print("\nSnowball Stemming:")
print([snowball.stem(word) for word in ws_tokens])
```

```python
lemmatizer = WordNetLemmatizer()

print("\nLemmatization:")
print([lemmatizer.lemmatize(word) for word in ws_tokens])
```
